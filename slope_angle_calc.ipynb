```python
theta = 0.1
def max_dbdz(Sinf, delta,gamma):
    num = theta*(1+Sinf**2)**(0.5)
    denom = Sinf**2*gamma*delta
    add = theta/(1-gamma)
    return ((num)/denom-add)**(-1) #, gamma
```

```python
Sinf = np.linspace(0.1,2,1000)
delta = np.linspace(0.01,1,1000)
gamma = np.linspace(0.01,0.99,1000)
S_inf, Gamma = np.meshgrid(Sinf,gamma)
S_inf, Delta = np.meshgrid(Sinf,delta)
# amplitude = max_dbdz(S_inf,Delta, Gamma)
```

```python
figure = plt.figure(figsize=(10,6))
lvl = np.linspace(1e-3,1000,91)
num = 0.8*S_inf**2*Gamma #+0.5*Gamma*S_inf**2
denom =  (1+S_inf**2)**(0.5)*(1-Gamma)
mu_a2 = (num/denom)
lvl = np.power(10,np.linspace(-5,4,91))  # Desired number of levels
plt.contourf(Sinf,gamma,np.abs(mu_a2),cmap="seismic",levels=lvl,norm=cl.LogNorm())
plt.plot(Sinf,(1+Sinf**2)**(-1),c="k")
plt.plot(Sinf,(3-Sinf**2)*(3*(1+Sinf**2))**(-1),c="k")
# plt.xlim([0,0.5])

plt.ylim([0.01,0.99])

plt.xlabel(r"$S_\infty$",fontsize=14)
plt.ylabel(r"$\gamma$",fontsize=14)

plt.colorbar(label=r"Ratio")
plt.savefig("ratio.jpg")
plt.show()
```
