```python
theta = 0.1
def max_dbdz(Sinf, delta,gamma):
    num = theta*(1+Sinf**2)**(0.5)
    denom = Sinf**2*gamma*delta
    add = theta/(1-gamma)
    return ((num)/denom-add)**(-1) #, gamma
```

```python
def slope_angle_over_time(t,g,S,d):
    num = -theta*g*(1-d*S**2*(1+S**2)**(-0.5)*np.cos(t))
    denom = 1 - g + d*S**2*g*(1+S**2)**(-0.5)*np.cos(t)
    return num/denom
```

```python
t = np.linspace(0,3*2*np.pi,150)

plt.plot(t/np.pi,slope_angle_over_time(t,1/(1+0.01),0.1,0.8))
```

```python
def avg_slope_angle(g,d,S):
    dist = np.sqrt((1-g)**2-(d*S**2*g*(1+S**2)**(-0.5))**(2)+0*1j)
    return theta*(dist-1)/dist
```

```python
def avg_minus_max(g,d,S):
    avg = avg_slope_angle(g,d,S)
    max_slope = -theta*g*(1+d*S**2/(1+S**2)**(0.5))/(1-g-d*g*S**2/(1+S**2)**(0.5))
    return avg-max_slope
```

```python
figure = plt.figure(figsize=(10,6))
lvl = np.linspace(0,1,101)
mu_a2 = (-theta*Gamma/(1-Gamma))/avg_slope_angle(Gamma,0.8,S_inf)
plt.contourf(Sinf,gamma, mu_a2.real,cmap="RdYlBu",levels=lvl,vmax=1,vmin=0,extend="both") # np.arctan(mu_a)/(0.5*np.pi)
plt.plot(Sinf,(1+Sinf**2)**(-1),c="k")
plt.plot(Sinf,(3-Sinf**2)*(3*(1+Sinf**2))**(-1),c="k")
# plt.xlim([0,0.5])

# plt.ylim([0,0.99])

plt.xlabel(r"$S_\infty$",fontsize=14)
plt.ylabel(r"$\gamma$",fontsize=14)

plt.colorbar(label=r"$\arctan(\hat{s}_{TOTAL})/0.5\pi$")
plt.savefig("total.jpg")
plt.show()
```
