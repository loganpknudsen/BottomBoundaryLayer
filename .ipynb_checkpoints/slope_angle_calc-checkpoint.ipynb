```python
theta = 0.1
def max_dbdz(Sinf, delta,gamma):
    num = theta*(1+Sinf**2)**(0.5)
    denom = Sinf**2*gamma*delta
    add = theta/(1-gamma)
    return ((num)/denom-add)**(-1) #, gamma
```

```python
Sinf = np.linspace(0.01,2,1000)
delta = np.linspace(0,1,1000)
gamma = np.linspace(0.01,0.999,1000)
S_inf, Gamma = np.meshgrid(Sinf,gamma)
S_inf, Delta = np.meshgrid(Sinf,delta)
# amplitude = max_dbdz(S_inf,Delta, Gamma)
```

```python
def slope_angle_over_time(t,g,S,d):
    num = -theta*g*(1-d*S**2*(1+S**2)**(-0.5)*np.cos(t))
    denom = 1 - g + d*S**2*g*(1+S**2)**(-0.5)*np.cos(t)
    return num/denom
```

```python
def strat_over_time(t,g,S,d):
    return  1 - g + d*S**2*g*(1+S**2)**(-0.5)*np.cos(t)
```

```python
def avg_slope_angle(g,d,S):
    dist = np.sqrt((1-g)**2 - d**2*S**4*g**2/(1+S**2)+0*1j)
    return theta*(1-1/dist)
```

```python
def mean_background_slope(g,S):
    return -theta*g/(1-g)
```

```python
t = np.linspace(0,2*np.pi+1/150,150)
S1 = 0.25
g1 = (1+S1**2)**(-1)
# g1 = (3-S1**2)*(3*(1+S1**2))**(-1)
d1 = 0.8
angle = slope_angle_over_time(t,g1,S1,d1) #/(0.5*np.pi)
avg = avg_slope_angle(g1,d1,S1) #/(0.5*np.pi)
plt.plot(t/(2*np.pi),angle,c="k")
plt.hlines(avg.real,xmin=t[0]/(2*np.pi),xmax=t[-1]/(2*np.pi),color="r",linestyle="--")
# plt.hlines(np.mean(slope_angle_over_time(t,g1,S1,d1)),xmin=t[0]/(2*np.pi),xmax=t[-1]/(2*np.pi),color="r",linestyle="--",alpha=0.75)
plt.xlim([0,1])
plt.xlabel(r"$tf^*/2\pi$")
plt.ylabel(r"$\hat{s}$")
plt.title(r"$S_\infty=$"+str(S1)+r" and $\gamma:=\gamma_u$")
plt.savefig("slope_over_time_w_avg_S_0_25.jpeg")
plt.show()
```

```python
def avg_minus_max(g,d,S):
    avg = avg_slope_angle(g,d,S)
    max_slope = -theta*g*(1+d*S**2/(1+S**2)**(0.5))/(1-g-d*g*S**2/(1+S**2)**(0.5))
    return avg-max_slope
```

```python
figure = plt.figure(figsize=(10,6))
lvl = np.linspace(0,1,101)
mu_a2 = (-theta*Gamma/(1-Gamma))/avg_slope_angle(Gamma,0.8,S_inf)
plt.contourf(Sinf,gamma, mu_a2.real,cmap="RdYlBu",levels=lvl,vmax=1,vmin=0,extend="both") # np.arctan(mu_a)/(0.5*np.pi)
plt.plot(Sinf,(1+Sinf**2)**(-1),c="k")
plt.plot(Sinf,(3-Sinf**2)*(3*(1+Sinf**2))**(-1),c="k")
# plt.xlim([0,0.5])

# plt.ylim([0,0.99])

plt.xlabel(r"$S_\infty$",fontsize=14)
plt.ylabel(r"$\gamma$",fontsize=14)

plt.colorbar(label=r"$\arctan(\hat{s}_{TOTAL})/0.5\pi$")
plt.savefig("total.jpg")
plt.show()
```
