```python
def convective_limit(S,d):
    num = np.sqrt(1+S**2)
    denom = np.sqrt(1+S**2)+d*S**2
    return num/denom
```

```python
def lower_limit(S):
    return (3-S**2)/(3*(1+S**2))
```

```python
conv_data_1 = data.where((data.strat_values<=convective_limit(data.slope_burger_number,data.delta[-2])),np.nan)
conv_data_1 = conv_data_1.where((np.abs(conv_data_1.frequency)>=0.4999) | (conv_data_1.strat_values>=lower_limit(conv_data_1.slope_burger_number)),np.nan)
conv_data_2 = data.where(data.strat_values<=convective_limit(data.slope_burger_number,data.delta[-1]),np.nan)
conv_data_2 = conv_data_2.where((np.abs(conv_data_2.frequency)>=0.4999) | (conv_data_2.strat_values>=lower_limit(conv_data_2.slope_burger_number)),np.nan)
```

```python
def m_t(t, m, fstar, delta, lmbd, phi):
    # print(m+delta*lmbd/(fstar)*np.sin(t+phi))
    return m+delta*lmbd/(fstar)*np.sin(t+phi)
```

```python
def PSI_system(v,t,m,theta,gm,S2,delta,lmbd,phi):
    psi, D = v
    fstar = f*np.cos(theta)*(1+S2)**(0.5)
    A0 = (np.tan(theta)**2-S2)/(np.tan(theta)*(1+S2))*(m-np.tan(theta)*(np.tan(theta)**2-S2)**(-1)*((S2/(np.tan(theta)**2))*(1-gm*np.cos(theta)**(-4))+np.tan(theta)**2))
    A1 = m+(S2-np.tan(theta)**2)*(np.tan(theta)*(1+S2))**(-1)
    dpsidt = (A0-A1*m_t(t,m,fstar,delta,lmbd,phi))*D
    dDdt = (1+m_t(t,m,fstar,delta,lmbd,phi)**2)**(-1)*psi
    return [dpsidt, dDdt]
```

```python
tau = 2*np.pi
dt = 200
phi = np.pi/2
# d_di = 12
f = 1e-4
theta = 0.1
print("check")
data_run = conv_data_1.isel(delta=-2)
print("check")
strt_ind = 25
j = 20
t = np.linspace(0, tau+1/dt, dt)
WSPRT = []
WSPxRT = []
WSPyRT = []
BFLUXRT = []
BFLUXxRT = []
BFLUXzRT = []
GSPRT = []
ke_l = []
q = 0
for j in range(len(data_run.slope_burger_number)):
    WSPRT3 = []
    WSPxRT3 = []
    WSPyRT3 = []
    BFLUXRT3 = []
    BFLUXxRT3 = []
    BFLUXzRT3 = []
    GSPRT3 = []
    ke_l3 = []
    if j%50 == 0:
        print(j)
    for i in range(len(data_run.strat_values.isel(slope_burger_number=j))):
        WSPRT2 = []
        WSPxRT2 = []
        WSPyRT2 = []
        BFLUXRT2 = []
        BFLUXxRT2 = []
        BFLUXzRT2 = []
        GSPRT2 = []
        ke_l2 = []
        S2 = data_run.slope_burger_number[j].values**2
        N2 = S2*f**2/theta**2
        for d_di in [0]:
            gm = data_run.strat_values.isel(slope_burger_number=j,strat_index=i).values
            if np.isnan(data_run.growth_rate.isel(slope_burger_number=j,strat_index=i).values): # data.growth_rate.isel(delta=d_di,slope_burger_number=j,strat_index=i)<1e-2 and 
                WSPRT2.append(np.nan)
                WSPxRT2.append(np.nan)
                WSPyRT2.append(np.nan)
                BFLUXRT2.append(np.nan)
                BFLUXxRT2.append(np.nan)
                BFLUXzRT2.append(np.nan)
                GSPRT2.append(np.nan)
                ke_l2.append(np.nan)
            else:
                delta = data_run.delta.values
                m_max = data_run.slope_angle.isel(slope_burger_number=j,strat_index=i).values
                lmbd = N2*theta*gm/f
                sol = sc.odeint(PSI_system, [0, 1], t, args=(m_max, theta, gm, S2, delta, lmbd, phi))
                n = (f*(1+S2)**(0.5)/lmbd)
                beta = (1+S2)**(0.5)
                alpha = N2*(1-gm)*theta/(f*lmbd)
                Ri = N2*(1-gm)/lmbd**2
                fstar = f*beta
                w = ((m_t(t, m_max, fstar, delta, lmbd, phi))**2+1)**(-1)*sol[:,0]
                u = m_t(t, m_max, fstar, delta, lmbd, phi)*w*n
                v = -f*(gm*S2/theta-theta+m_max)*sol[:,1]/(lmbd)
                b = -N2*theta*(m_max+(1-gm)/theta)*sol[:,1]/(N2*(1-gm))
                ke = 0.5*(u**2+v**2+n**2*w**2)
                BFLUX = alpha*beta**(-1)*u*b+w*b*Ri
                BFLUX_int = sc.simpson(BFLUX,x=t)
                BFLUXx = alpha*beta**(-1)*u*b
                BFLUXx_int = sc.simpson(BFLUXx,x=t)
                BFLUXz = w*b*Ri
                BFLUXz_int = sc.simpson(BFLUXz,x=t)
                GSP = -v*w
                GSP_int = sc.simpson(GSP,x=t)
                WSP = -u*w*delta*np.cos(t+phi)+v*w*(f*delta/fstar*np.sin(t+phi))
                WSP_int = sc.simpson(WSP,x=t)
                WSPx = -u*w*delta*np.cos(t+phi)
                WSPx_int = sc.simpson(WSPx,x=t)
                WSPy = v*w*(f*delta/fstar*np.sin(t+phi))
                WSPy_int = sc.simpson(WSPy,x=t)
                # ke = BFLUX_int+GSP_int+WSP_int
                WSPRT2.append((WSP_int)/(ke[-1]-ke[0]))
                WSPxRT2.append((WSPx_int)/(ke[-1]-ke[0]))
                WSPyRT2.append((WSPy_int)/(ke[-1]-ke[0]))
                BFLUXRT2.append((BFLUX_int)/(ke[-1]-ke[0]))
                BFLUXxRT2.append((BFLUXx_int)/(ke[-1]-ke[0]))
                BFLUXzRT2.append((BFLUXz_int)/(ke[-1]-ke[0]))
                GSPRT2.append((GSP_int)/(ke[-1]-ke[0]))
                ke_l2.append(ke[-1]-ke[0])
        WSPRT3.append(WSPRT2)
        WSPxRT3.append(WSPxRT2)
        WSPyRT3.append(WSPyRT2)
        BFLUXRT3.append(BFLUXRT2)
        BFLUXxRT3.append(BFLUXxRT2)
        BFLUXzRT3.append(BFLUXzRT2)
        GSPRT3.append(GSPRT2)
        ke_l3.append(ke_l2)
    WSPRT.append(WSPRT3)
    WSPxRT.append(WSPxRT3)
    WSPyRT.append(WSPyRT3)
    BFLUXRT.append(BFLUXRT3)
    BFLUXxRT.append(BFLUXxRT3)
    BFLUXzRT.append(BFLUXzRT3)
    GSPRT.append(GSPRT3)
    ke_l.append(ke_l3)
WSPRT = np.array(WSPRT)
WSPxRT = np.array(WSPxRT)
WSPyRT = np.array(WSPyRT)
BFLUXRT = np.array(BFLUXRT)
BFLUXxRT = np.array(BFLUXxRT)
BFLUXzRT = np.array(BFLUXzRT)
GSPRT = np.array(GSPRT)
ke_l = np.array(ke_l)
```

```python
def avg_slope_angle(g,d,S):
    theta = 0.1
    dist = np.sqrt((1-g)**2 - d**2*S**4*g**2/(1+S**2)+0*1j)
    return theta*(1-1/dist)
```
