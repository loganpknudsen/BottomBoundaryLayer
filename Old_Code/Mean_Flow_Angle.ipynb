```python
import numpy as np
import scipy as sp
import matplotlib.pyplot as plt
```

```python
S2 = 1.5**2
gml = (1+(1-10/6)*S2)/(1+S2) 
gmu = (1+(1-10/11)*S2)/(1+S2)
dgm = 100
theta = 0.1
gm_list = np.linspace(gml,gmu,dgm)
deltas = np.linspace(0,1,100)
mean_angle = (-1*(1-gm_list)/theta)*np.ones((100,100))
Ri_inv = np.linspace(0.6,1.1,100)
```

```python
plt.contourf(Ri_inv,deltas,mean_angle)
plt.colorbar()
```
